# Exploratory Data Analysis (EDA)



```python
import matplotlib.pyplot as plt
```

```python
# Employee Attrition Count

df['Attrition'].value_counts().plot(kind='bar')

plt.title("Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Count")
plt.show()
```

```python
# Gender Distribution

df['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Gender Distribution")
plt.show()
```

```python
# Department Distribution

df['Department'].value_counts().plot(kind='bar')
plt.show()
```

```python
# Monthly Income Distribution

plt.hist(df['MonthlyIncome'], bins=20)
plt.title("Monthly Income")
plt.show()
```

```python
# Age Distribution

plt.hist(df['Age'])
plt.show()
```

```python
# Attrition by Gender

pd.crosstab(df['Gender'],df['Attrition']).plot(kind='bar')
plt.show()
```

```python
# Attrition by Department

pd.crosstab(df['Department'],df['Attrition']).plot(kind='bar')
plt.show()
```

```python
# Attrition by Job Role

pd.crosstab(df['JobRole'],df['Attrition']).plot(kind='bar', figsize=(12,5))
plt.show()
```

```python
# Attrition by Overtime

pd.crosstab(df['OverTime'],df['Attrition']).plot(kind='bar')
plt.show()
```

```python
# Monthly Income vs Attrition

import matplotlib.pyplot as plt

plt.boxplot([
df[df['Attrition']=='Yes']['MonthlyIncome'],
df[df['Attrition']=='No']['MonthlyIncome']
])

plt.xticks([1,2],['Yes','No'])
plt.show()
```
